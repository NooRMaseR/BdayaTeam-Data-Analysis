    <li>متوسط الكميات: ما هي فئة المنتج التي يشتري منها العملاء أعلى "متوسط" للكمية (Quantity) في الفاتورة الواحدة؟</li>
    <li>عدد الفواتير: كم عدد الفواتير (استخدم count لعمود Transaction ID) التي قام بها الرجال مقارنة بالنساء؟</li>

        تقرير الإدارة الشامل (سؤال التميز): قم بإنشاء جدول محوري يعرض إجمالي المبيعات (Total Amount) لكل فئة منتج، مقسماً على الجنس، مع إضافة عمود وصف "الإجمالي الكلي" (margins=True).
    </li>

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv(
    "retail_sales_dataset.csv",
    dtype={
        "Transaction ID": np.uint16,
        "Age": np.uint8,
        "Quantity": np.uint8,
        "Price per Unit": np.uint16,
        "Total Amount": np.uint16,
    },
    parse_dates=["Date"],
)
df
```

```python
df.groupby("Product Category")[['Total Amount']].sum().sort_values("Total Amount", ascending=False)
```

```python
pd.pivot_table(
    data=df,
    values="Total Amount",
    index="Product Category",
    aggfunc='sum',
).sort_values("Total Amount", ascending=False)
```

```python
pd.pivot_table(
    data=df,
    values="Total Amount",
    index="Product Category",
    columns="Gender",
    aggfunc="sum"
)
```

```python
pd.pivot_table(
    data=df,
    index='Product Category',
    values='Quantity'
)
```

```python
pd.pivot_table(
    data=df,
    values="Transaction ID",
    index="Gender",
    aggfunc='count'
)
```

```python
pd.pivot_table(
    data=df,
    values="Total Amount",
    index="Gender",
    columns="Product Category",
    aggfunc='sum',
    margins=True,
    margins_name="Total"
)
```
